```python
# ==========================================
# STEP 2: DOWNLOAD THE DATASETS (BUG FIX FOR PYTHON 3.13)
# ==========================================
import os
import kagglehub
from google.colab import userdata

# Retrieve the token from secrets and strip out any hidden newline characters (\r\n)
raw_token = userdata.get("KAGGLE_API_TOKEN")
clean_token = raw_token.strip() if raw_token else ""

# Set the environment variable with the clean token
os.environ["KAGGLE_API_TOKEN"] = clean_token


# 2. Download RoEduNet-SIMARGL2021
print("📥 Downloading RoEduNet-SIMARGL2021...")
simargl_path = kagglehub.dataset_download("h2020simargl/simargl2022")
print(f"✅ RoEduNet-SIMARGL2022 downloaded to: {simargl_path}")
```

```python
# ==========================================
# STEP 2: RE-LINK THE 'sensor_db' SHORTCUT
# ==========================================
import os

# Define the correct path to the 2022 dataset we just downloaded
simargl2022_path = "/root/.cache/kagglehub/datasets/h2020simargl/simargl2022/versions/1"

# Delete the old shortcut if it exists
if os.path.exists("sensor_db"):
    os.remove("sensor_db")
    print("Removing old 'sensor_db' shortcut...")

# Create the new shortcut pointing to the 2022 folder
os.symlink(simargl2022_path, "sensor_db")
print("✅ Successfully linked 'sensor_db' to your 2022 Kaggle dataset!")
```
